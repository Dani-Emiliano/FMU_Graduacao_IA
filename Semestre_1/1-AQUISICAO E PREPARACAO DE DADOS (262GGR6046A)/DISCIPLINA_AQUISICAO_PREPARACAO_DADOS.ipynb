## 1. Dados Abertos: Definição e Características

**Definição:** Dados que podem ser acessados, utilizados, modificados e compartilhados por qualquer pessoa para diferentes finalidades.

### 3 Características Principais

1. **Acesso e Disponibilidade**
   - Disponibilizados por completo
   - Formatos que facilitem modificação
   - Sem cobranças
   - Normalmente em portais governamentais

2. **Reutilização**
   - Permitem reuso em outras bases de dados
   - Sem restrições comerciais

3. **Participação Universal**
   - Qualquer pessoa pode usar e distribuir
   - Sem restrições discriminatórias

### 3 Leis de David Eaves (Ativista de Dados Abertos)

1. **Se não pode ser encontrado e indexado na Web, não existe**
   - Dados ocultos = dados inacessíveis

2. **Se não estiver aberto em formato compreensível por máquina, não pode ser reaproveitado**
   - Necessário formato estruturado (CSV, XML, RDF, não PDF)
   - Máquinas precisam processar dados

3. **Se algum dispositivo legal não permitir replicação, não é útil**
   - Licenças restritivas limitam uso
   - Dados com restrições legais = menos valor

---

## 2. Oito Princípios dos Dados Abertos Governamentais (Califórnia, 2007)

| Princípio | Descrição |
|-----------|----------|
| **Completos** | Sem controle de privacidade/acesso/segurança; abertos para qualquer uso |
| **Primários** | Publicados em máximo detalhe, não sumarizados/agrupados |
| **Atuais** | Publicados rápido para não perder valor; atualizados conforme necessário |
| **Acessíveis** | Disponibilizados para maior público; normalmente na internet |
| **Processáveis por Máquina** | Estrutura que facilite processamento automatizado |
| **Acesso Não Discriminatório** | Disponíveis a qualquer pessoa, sem registro |
| **Formatos Não Proprietários** | Em formatos abertos, não proprietários |
| **Licenças Livres** | Não sujeitos a copyright/patentes/marcas |

## 1. Definição ETL

**ETL (Extract, Transform and Load) = Extração, Transformação e Carga**

- Procedimentos responsáveis pela extração de dados de várias fontes
- Limpeza, otimização e inserção em Data Warehouse
- Uma das fases mais críticas na construção de DW
- Grandes volumes processados nesta fase

### Fontes Heterogêneas
- Bancos de dados relacionais (maioria)
- Bases de dados Access
- Planilhas (Excel)
- Arquivos CSV
- Arquivos XML
- Sistema comunica com heterogeneidade (formato e normalização)

---

## 2. Três Fases do ETL

### FASE 1: EXTRAÇÃO (Extract)

- Coleta de dados dos sistemas de origem (Data Sources/sistemas operacionais)
- Extração e transferência para staging area
- **ETL opera independente dos sistemas operacionais**
- Múltiplos servidores e tecnologias podem ser lidos

### FASE 2: TRANSFORMAÇÃO (Transform)

Após extração, **tratar, limpar e filtrar** para garantir integridade.

**Atividades principais:**
- Limpeza (erros digitação, caracteres desconhecidos, abreviações)
- Descoberta violação integridade
- **Padronização de valores**
- Identificação anomalias
- Resolução de conflitos **antes** inserir no DW

**Conflitos Semânticos:**
- Mesmo nome para diferentes entidades
- Diferentes nomes para mesma entidade

**Conflitos Estruturais:**
- Diferenças unidades (cm vs polegadas)
- Diferenças precisão (0,12 vs 0,123456)
- Diferenças códigos/expressões (M/F vs 0/1)
- Diferenças granularidade (semana vs mês)
- Diferenças abstração (1 atributo vs rua+complemento)

### FASE 3: CARGA (Load)

- Estruturar e carregar no Data Warehouse
- Modelo dimensional
- Varia conforme necessidades da organização
- Latência e reposição = opções estratégicas

---

## 3. Arquitetura ETL Completa

```
Fontes → Extração → Staging Area → Transformação → Data Warehouse → OLAP
                      (área provisória)   (limpeza)
```

---

## 4. OLAP — Online Analytical Processing

- Ferramenta **após DW construído**
- Acesso e análise dos dados
- Análises **multidimensionais**
- Cálculos complexos, tendências, modelação

---

## 5. Dez Vantagens do ETL

✅ **Qualidade dados** — algoritmos complexos detectam erros  
✅ **Separação funcionalidade/dados** — sistemas operacionais não afetam DW  
✅ **Desenvolvimento rápido** — framework pré-construído  
✅ **Manutenção simples** — alterações em um lugar  
✅ **Metadados automáticos** — documentação gerada  
✅ **Desempenho otimizado** — índices e cache  
✅ **Portabilidade** — funciona em diferentes plataformas  
✅ **Múltiplas conexões** — transparente ao usuário  
✅ **Reinicialização** — recuperação em caso de falha  
✅ **Segurança modular** — controle de acesso granular

## 1. O que é Normalização?

**Normalização:** Transformação de variáveis de forma **independente** para uma **escala comum**, sem distorcer as diferenças no intervalo de valores.

**Objetivo:** Transformar dados em diferentes grandezas/escalas para conjunto padronizado.

### Distinções Importantes

- **Padronização:** Criar distribuição normal (média=0, desvio-padrão=1)
- **Redimensionamento:** Mudar unidade por constante (°C → °F)
- **Normalização:** Reescalonar para intervalo [0,1] sem distorcer proporções

---

## 2. Quatro Tipos de Normalização

### 1. MÍNIMO-MÁXIMO

**Fórmula:** `(x - min) / (max - min)`  
**Intervalo:** [0, 1]  
**Sensibilidade outliers:** ⚠️ **ALTA**  
**Função Python:** `preprocessing.MinMaxScaler()`

```python
from sklearn import preprocessing
scaler_mm = preprocessing.MinMaxScaler(feature_range=(0, 1))
dados_mm = scaler_mm.fit_transform(dados)
```

**Quando usar:** Sem outliers + intervalo conhecido

---

### 2. Z-SCORE (Padronização)

**Fórmula:** `(x - média) / desvio-padrão`  
**Intervalo:** [-∞, +∞] (números negativos e positivos)  
**Sensibilidade outliers:** ✅ **BAIXA** (robusto)  
**Função Python:** `preprocessing.StandardScaler()`

```python
from sklearn import preprocessing
scaler_z = preprocessing.StandardScaler()
dados_z = scaler_z.fit_transform(dados)
```

**Resultado:** Média = 0, Desvio-padrão = 1  
**Quando usar:** Distribuição normal  

---

### 3. ESCALA DECIMAL

**Fórmula:** `x / 10^m` (m torna máximo < 1)  
**Intervalo:** Variável  
**Sensibilidade outliers:** ⚠️ **MUITO ALTA**  
**Função Python:** Manual (não há pronta)  

```python
m = len(str(int(max(dados))))
dados_decimal = dados / (10 ** m)
```

---

### 4. ROBUST SCALER (Para Outliers)

**Fórmula:** `(x - mediana) / IQR`  
**IQR:** Q3 - Q1 (intervalo interquartil)  
**Intervalo:** [-∞, +∞]  
**Sensibilidade outliers:** ✅✅ **MUITO BAIXA**  
**Função Python:** `preprocessing.RobustScaler()`

```python
from sklearn import preprocessing
scaler_robust = preprocessing.RobustScaler()
dados_robust = scaler_robust.fit_transform(dados)
```

**Quando usar:** Muitos outliers nos dados

---

## 3. Comparativo dos Quatro Métodos

| Método | Intervalo | Outliers | Função | Caso |Uso |
|--------|-----------|----------|--------|------|-----|
| **Mínimo-Máximo** | [0,1] | ⚠️ Sensível | MinMaxScaler | Sem outliers, intervalo conhecido | Recomendado para dados limpos |
| **Z-Score** | [-∞,+∞] | ✅ Robusto | StandardScaler | Distribuição normal | **Mais utilizado** |
| **Escala Decimal** | Variável | ⚠️ Muito sensível | Manual | Educacional | Raro em produção |
| **Robust Scaler** | [-∞,+∞] | ✅✅ Muito robusto | RobustScaler | Muitos outliers | Dados com anomalias |

## 1. Pirâmide Dados → Informação → Conhecimento → Sabedoria

```
           SABEDORIA
        (aplicação em decisão)
              ↑
       CONHECIMENTO
    (relacionamento de info)
              ↑
       INFORMAÇÃO
      (dados processados)
              ↑
         DADOS
        (brutos)
```

**Transformação:** Dados brutos → Transformação → Informação → Conhecimento → Sabedoria

---

## 2. Análise Exploratória de Dados (AED)

**Definição:** A **principal etapa da análise dos dados**, onde deve-se:
- Organizar dados
- Resumir informações
- Aplicar cálculos
- Resumir em formato visual adequado

### Diferença: AED vs Outras Abordagens

| Abordagem | Estratégia |
|-----------|----------|
| **Estatística Clássica** | Problema → Dados → **Modelo** → Análise |
| **Estatística Bayesiana** | Problema → Dados → Modelo Priori → Análise |
| **AED** | Problema → Dados → **Análise** → Modelo |

**Insight:** AED **explora dados ANTES de definir modelo** (data-first)

---

## 3. Classificação de Dados em AED

### Valores Numéricos (Quantitativos)

**Discretos:** Números inteiros  
- Exemplo: idade (18, 19, 20)

**Contínuos:** Números reais em intervalo  
- Exemplo: peso (75,5 kg, 60,3 kg, 80,7 kg)

### Valores Não-Numéricos (Qualitativos)

**Nominais:** Sem ordem natural  
- Exemplos: nome, cor, raça, sexo

**Ordinais:** Com ordem natural  
- Exemplos: classe social (baixa/média/alta), tamanho roupa (P/M/G)

---

## 4. Etapas da AED

### 4.1 COLETA E ORGANIZAÇÃO
- Definir métodos de coleta
- Estabelecer objetivos da pesquisa
- Unificar dados de diferentes fontes
- Considerar natureza dos dados

### 4.2 TRATAMENTO DE DADOS

#### Missing Values (Dados Ausentes)
**Soluções:**
- Descartar registro com dados ausentes
- **Imputação:** Preencher com:
  - Média ou mediana
  - Repetição do último valor
  - Regressão linear

**Crítico:** Técnica depende **do tipo do dado faltante**

#### Normalização
Transformar valores em escala comum [0,1] sem distorcer diferenças.

#### Outliers (Valores Extremos)
**Definição:** Valores que fogem do padrão observado

**Causas:**
- Entrada errada de dados
- Erros em tempo de tratamento

**Tratamento:**
- Remoção
- Investigação individual
- Transformação

**Decisão Crítica:** Depende do contexto
- Em análise de **fraude**: outlier = insight valioso (anomalia buscada)
- Em análise de **comportamento**: outlier = distorção (deve remover)

### 4.3 ANÁLISE DE DADOS

#### Análise Univariada
Avalia **cada variável isoladamente**

Objetivo: Frequência de valores possíveis

**Exemplo:**
```
Variável "idade" com valores {18, 19, 18, 21, 21, 18}
Frequência: {18 → 3, 19 → 1, 21 → 2}
```

#### Análise Multivariada
Estabelece **relações entre 2+ variáveis**

Exemplo: altura da pessoa mediante sexo e idade

#### Correlações
Dois acontecimentos **tendem a ocorrer sincronizados**

(Não são necessariamente causais)

**Exemplos:**
- Inflação alta ↔ desemprego aumenta
- Cliente compra pão todo dia ↔ satisfação com pão alta

### 4.4 APRESENTAÇÃO E INTERPRETAÇÃO

#### Formas de Apresentação

**Tabular:**
- Tabela com título, títulos colunas, corpo dados
- Precisa, mas menos intuitiva

**Gráfica:**
- Deve ser auto-explicativa e fácil compreensão
- Visualmente mais intuitiva
- Tipos: colunas, barras, linhas, setores, radar, etc.

**Importância:** Escolher tipo adequado ao tipo de dados
- Quantitativos → gráficos de linhas, colunas
- Qualitativos nominais → gráficos de barras, setores
- Qualitativos ordinais → gráficos de barras ordenadas

**Exemplo de erro:**
- Usar gráfico radar para estado civil (nominal)
- Não deixa clara a correlação entre valores e frequência

## 6. Pré-processamento: Reescalonar vs Normalizar vs Padronizar

### Problema
Características em escalas diferentes (altura em cm, peso em gramas) podem dar importância errada ao algoritmo ML.

### Soluções

**Reescalonar:** Mudar unidade de medida (ex: cm → m)  
**Normalização:** (x - min) / (max - min) → intervalo [0,1]  
**Padronização:** (x - média) / desvio-padrão → média=0, σ=1

✅ Reproduzível  

### Problema Principal
Você removia dados que poderiam ser **insights valiosos** (fraude, anomalia de interesse).

---

## 2. ETL + ML Moderno (Cascata de Modelos)

```
Coleta → Extração → TRANSFORMAÇÃO (Híbrida) → Load → ANÁLISE COM ML
                    ├─ Normalização básica (escala)
                    ├─ Imputação (média/mediana)
                    ├─ ML modelo 1: Detecção de outliers
                    ├─ Tratamento contextual (decide remover/manter)
                    └─ Validação com ML
                                           ↓
                                    ML modelo 2: Análise real
```

### Sua Arquitetura
✅ **2 modelos em sequência:** primeiro detecta, depois analisa  
✅ Outliers identificados **com contexto do negócio**  
✅ Aproveita aprendizado supervisionado  
✅ Adaptável por domínio  
⚠️ Mais complexo (mais pontos de falha)

---

## 3. Três Cenários Onde Funciona

### Cenário 1: Sua Caso — Compatibilidade Física com Padrão de Marca

```
Dataset: Modelos potenciais (altura, peso, proporções)

PIPELINE:
1. Extração bruta
2. Normalização básica (escalas diferentes)
3. ML Modelo 1 (Unsupervised): Detectar outliers
   └─ Identifica "altura = 2.50m" ou "peso = 250kg"
   
4. Decisão contextual:
   ✅ MANTER se: "Procuramos pessoas fora do padrão"
   ❌ REMOVER se: "Apenas padrão comercial"
   
5. ML Modelo 2 (Supervisionado): Classificar compatibilidade
   └─ Treina com exemplos de sucesso (modelos aprovados vs rejeitados)
   └─ Aprende que outliers tratados = mais acurácia
```

**Vantagem:** O outlier não é "erro", é **variação relevante** que o modelo aprende a interpretar.

### Cenário 2: Detecção de Fraude em Transações

```
Dataset: Histórico transacional

1. Normalização: Valores em escala comum
2. ML Modelo 1 (Detecção): Isola transações anômalas
   └─ Um saque de R$50k em conta que nunca saca
   
3. Decisão contextual:
   ✅ INVESTIGAR (não remover!)
   ❌ Remover mataria o insight mais valioso
   
4. ML Modelo 2 (Classificação): "Fraude vs Legítimo"
   └─ Aprende que anômalias são **mais informativas** que padrão
```

**Insight:** Os outliers são exatamente o que você **quer prever**.

### Cenário 3: Análise de Desempenho com Exceções

```
Dataset: Performance de vendedores

1. Normalização: Métricas em escala comparável
2. ML Modelo 1 (Clustering): Identifica "top performers"
   └─ Vendedor A: 500% acima da média
   
3. Decisão:
   ✅ SEPARAR (investigar por quê é diferente)
   └─ Pode gerar insights: tática, mercado, atuação
   
4. ML Modelo 2 (Regressão): Prever performance
   └─ Aprende que exceções revelam **padrões ocultos**
```

## 4. Aprendizado Supervisionado na Detecção

Você pode **rotular outliers manualmente** e treinar um classificador:

```python
import numpy as np
from sklearn import preprocessing

# Exemplo conceitual
dados = np.array([
    [1.65, 10],   # normal
    [1.80, 20],   # normal
    [1.72, 30],   # normal
    [2.50, 250]   # outlier (altura muito alta, peso muito alto)
])

eh_outlier = np.array([0, 0, 0, 1])  # Você marca

# ML Modelo 1: Aprende padrão de outliers
from sklearn.ensemble import IsolationForest
classifier = IsolationForest(contamination=0.25)
outliers_detectados = classifier.fit_predict(dados)

# ML Modelo 2: Usa dataset ANOTADO
from sklearn.ensemble import RandomForestClassifier
modelo_final = RandomForestClassifier()
modelo_final.fit(dados[eh_outlier == 0], y_treino)  # Treina com dados limpos
```

## 6. Conclusão

**Sua abordagem está correta:**

1. **Normalização** (escala comum) → **Sempre faz no T da ETL**
   - Necessária para qualquer coisa depois

2. **Detecção de outliers com ML** → **Modelo 1 (após Load ou antes)**
   - Usa aprendizado supervisionado/não-supervisionado
   - Marca o dado, não necessariamente remove

3. **Análise/predição real** → **Modelo 2 (após normalização + marcação)**
   - Treina com dados anotados
   - Aprende a lidar com outliers no contexto
